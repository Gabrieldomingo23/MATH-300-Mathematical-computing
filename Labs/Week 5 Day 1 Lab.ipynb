### Problem 1
Write your own function performs vectors addition on lists instead of arrays.

```python
import numpy as np
def Matrixadd(x, y):
    ContList = []
        
    for i in range(len(x)):
            ContList.append(x[i]+y[i])
    return ContList

test1 = [4, 5, 8, 9]
test2 = [3, 8, 7, 10]
result = Matrixadd(test1, test2)
result
```

### Problem 3
Make a 2 by 2 invertible matrix, A.
1. Compute its eigenvectors and plot them using matplotlib
2. Compute A times the eigenvectors and plot them on the same plot.
3. What do you notice?

```python
import matplotlib.pyplot as plt
Matrix = np.array([4, 8, 9, 6])
Matrix = Matrix.reshape(2,2)
Inverse = np.linalg.inv(Matrix)

Eigen = np.linalg.eig(Matrix)[1]
Eigen
EigenA = Eigen @ Matrix

x1 = [Eigen[0][0], Eigen[0][1]]
      
y1 = [Eigen[1][0], Eigen[1][1]]
x2 = [EigenA[0][0], EigenA[0][1]]
y2 = [EigenA[1][0], EigenA[1][1]]
plt.scatter(x1, y1, label = 'Eigen Vectors', color = 'grey')
plt.scatter(x2, y2, label = 'Eigen * A',color = 'red')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.title("Eigenvectors compared to A*Eigen")
plt.plot([0, x1[0]], [0, y1[0]], color = 'grey')
plt.plot([0, x1[1]], [0, y1[1]], color = 'gray')
plt.plot([0, x2[0]], [0, y2[0]], color = 'red')
plt.plot([0, x2[1]], [0, y2[1]], color = 'red')
plt.grid('on')
# Changes the angle between the eigen vectors
```

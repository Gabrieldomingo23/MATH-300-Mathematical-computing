### Problem 5
Write a recursive function `fibonacci(n)` that computes the $n$-th Fibonacci number. Compute the time complexity of your `fibonacci(n)` function.

````python
def fibonacci(n):
    print(n)
    if(n == 0):
        return n
    elif(n == 1):
        return 1 + fibonacci(n-1)
    elif(n == 2):
        return 1 + fibonacci(n-1)
        
    else:
        return fibonacci(n-1) + n
print(fibonacci(4))
0, 1, 1, 2, 3, 5, 8, 
````

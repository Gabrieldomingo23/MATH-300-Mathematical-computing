### Problem 1
Plot the function $f(x)=x^2$ for $x\in [-1,1]$ using `np.linspace()` with 10, 100, 1000 slices. Make sure your plot is as informative as possible.

```python
import pandas as pd
import geopandas as gdp
import matplotlib.pyplot as plt
import math
import numpy as np
import sympy as sp

def f(x):
    return x**2

h = np.linspace(-1, 1, 10)
m = np.linspace(-1, 1, 100)
n = np.linspace(-1, 1, 1000)


fig, axes = plt.subplots(1,3, sharey = True)
axes[0].scatter(h, f(h), color = "red")
axes[1].scatter(m, f(m), color = "green")
axes[2].scatter(n, f(n), color = "blue")
axes[0].set_title("10")
axes[1].set_title("100")
axes[2].set_title("1000")
```

### Problem 2
Pick a function $f(x)$ you like and plot some values using `plt.scatter()`. Make sure your plot is as informative as possible.

```python
def f(x):
    return np.log(x)

x = np.linspace(0.1, 100, 10000)

plt.scatter(x, f(x), label = "Log(x)", color = "purple")
plt.title("log(x)")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
```

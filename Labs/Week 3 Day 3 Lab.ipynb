1. Create your own dataframe with at least 3 columns and 4 rows.
2. Save the dataframe to a csv file.
3. Read the csv back in.
4. Plot the data.

````python
import pandas as pd
import matplotlib.pyplot as plt

Name = ["Gabriel", "Cera", "Cody", "Michael"]
Birthday = ["09/16", "07/29", "07/08", "08/14"]
Order = [3, 2, 1, 4]

dataframe = pd.DataFrame({
    'Name' : Name,
    'Birthday' : Birthday,
    'Order' : Order 
})

dataframe
dataframe.to_csv("Siblings.csv", index = False)
````

````python
pd.read_csv("Siblings.csv")
plt.plot(dataframe['Name'], dataframe['Order'])
plt.title("Sibling Order")
plt.xlabel("Sibling")
plt.ylabel("Order")
````

### Problem 2
We want to compute $\int_0^1 x^2dx$. To do so, generate random points between 0 and 1. Accept them if they are below and reject them if they are above. Compute the average of the accepted values and plot the accepted vs rejected points.

```python
def f(x):
    return x**2
    
def function(n):
    x = np.random.uniform(0,1,n)
    y = np.random.uniform(0,1,n)
    inside =  y <= x**2
    rejected = y > x**2

    plt.scatter(x[inside], y[inside], s=5, label="accepted", color = 'blue')
    plt.scatter(x[rejected], y[rejected], s=5, label='rejected', color = 'red')

    h = np.linspace(0,1, 100000)

    plt.scatter(h, f(h), color = 'green', label = 'x^2')
    plt.legend()
    print("the average is", np.mean(x[inside]))


    
    
```

```python
function(100000)
```

```python
# the real value
import sympy as sp
x = sp.Symbol('x')

h = x**2

result = sp.integrate(h, (x, 0, 1))
print("The actual answer is", result)
```

### Problem 3 
Write a function that will do Problem 2 for any function and bounds.

```python
def general(f, n, lower, upper):
    x = np.random.uniform(lower,upper,n)
    y = np.random.uniform(lower,f(upper),n)
    inside =  y < f(x)
    rejected = y >= f(x)

    plt.scatter(x[inside], y[inside], s=5, label="accepted", color = 'blue')
    plt.scatter(x[rejected], y[rejected], s=5, label='rejected', color = 'red')

    h = np.linspace(lower,upper, 100000)

    plt.scatter(h, f(h), color = 'green', label = 'function')
    print("the average is", np.mean(inside))
    print("The answer is", (len(x[inside])/n) * (float(upper-lower) * f(upper)))
    plt.legend()

def f(x):
    return x**3 + 5*x**2 + 7
general(f, 100000, -2, 5)
```

```python
h = x**3+5*x**2+7

result = sp.integrate(h, (x, -2, 5))
print("The actual answer is", result)
```

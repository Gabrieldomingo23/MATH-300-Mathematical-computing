## Problem 2

1. Write a recursive function `factorial()` that takes in a natural number `n` and computes $n!=n(n-1)(n-2)\cdots3\cdot2\cdot 1$. 
2. Use `%timeit` to determine how fast your function calculates `factorial(150)`. 
3. Use `%timeit` to calculate how fast `math.factorial()` calculates the same quantity.
4. Were there any differences between the two methods?

```python
%timeit math.factorial(150) #1.61 microseconds += 26.3 ns per loop

def factorial(n):
    if n == 1:
        return n
    else:
        return n * factorial(n-1)

#%timeit factorial(150) #19.3 microseconds +- 156 ns per loop

#4 doing a recursive functino is slower than the math function
```

Recall for a quadratic equation of the form
$$ax^2 + bx + c = 0
$$

The solutions are given by the <i>quadratic formula</i>:

$$
x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}
$$

1. Write a function `quadratic()` that calculates the two roots given by the quadratic formula only if they are real.

3. Use `%timeit` to calculate how fast `sp.solve()` calculates the same quantities.
4. Were there any differences between the two methods?

```python

def quadratic(a,b,c):
    x1 = ((-1)*b + math.sqrt(b**2 - 4*a*c))/(2*a)
    x2 = ((-1)*b - math.sqrt(b**2 - 4*a*c))/(2*a)
    if (int(x1) == x1) & (int(x2) == x2):
        print(x1, x2)
    return
#%timeit quadratic(437, 915, 478) #722 ns +- 12.5 ns per loop wayyyyy slower
a = 437
b = 915
c = 478
quadratic(a, b, c)

x = sp.symbols('x')
#%timeit sp.solve((-b + math.sqrt(b**2 - 4 *a*c))/(2*a)) #136 microseconds +- 1.86 microseconds per loop
```

## Problem 4

For each of the following functions, use `sympy` calculate the first derivative, the second derivative, and the integral (all with respect to $x$):

1. $f(x)=x$
2. $f(x)=6x^{-1/2}$
3. $f(x)=\pi\sqrt{x}+x^5+7$
4. $f(x)=x(2x-5)(3+\tan(x))$

```python
x = sp.symbols('x')
f = x
g = 6 * (x**-.5)
h = math.pi*(x**.5) + x**5 + 7
i = x * (2*x -5)*(3 + sp.sin(x))
print(sp.diff(f, x))
print(sp.diff(sp.diff(f,x), x))
print(sp.integrate(f, x))
print(sp.diff(g, x))
print(sp.diff(sp.diff(g,x), x))
print(sp.integrate(g, x))
print(sp.diff(h, x))
print(sp.diff(sp.diff(h,x), x))
print(sp.integrate(h, x))
print(sp.diff(i, x))
print(sp.diff(sp.diff(i,x), x))
print(sp.integrate(sp.expand(i), x))
```

### Problem 1
Suppose $f(x)=x^2+6x+10$

1. Plot the function to visualize the minimum
2. Find the minimum using `scipy`

````python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
import math
from scipy.optimize import minimize



x = np.linspace(-10, 50, 500)
f = x**2 + 6*x + 10
plt.figure(figsize= (8,5))
plt.plot(x, f)
plt.xlabel("X-axis")
plt.ylabel('Y-axis')
plt.title("Function")
plt.grid()
plt.show()

````

````python
def g(x):
    return (x**2 + 6*x + 10)

result = minimize(g, x0=10)
print(result.x)
plt.plot(x, g(x))
plt.scatter(result.x[0], result.x)
````


## Problem 2

Find the maximum of $50x-x^2$.

````python
def h(x):
    return -(50*x - (x**2))
result = minimize(h, x0=10)
best_x = result.x[0]
maximum = -h(best_x)

print(best_x)
print(maximum)
plt.figure(figsize=(5, 8))
plt.plot(x, -1*h(x))
plt.scatter(best_x, maximum, color = 'red')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Function')
plt.grid()
plt.show()
````

### Problem 4
Design a function with multiple minima and optimize.

````python
def f(x):
    return ((1/12)*x**4 + x**3 + (5/2)*x**2)
x = np.linspace(-9, 3, 500)
result = minimize(f, x0=-6)
print(result.x[0])
print(f(result.x[0]))
plt.figure(figsize = (8,8))
plt.plot(x, f(x), label = 'function')
plt.scatter(result.x[0], f(result.x[0]), color = 'red', label = "minimum")
plt.legend()
plt.grid()
plt.show()
````

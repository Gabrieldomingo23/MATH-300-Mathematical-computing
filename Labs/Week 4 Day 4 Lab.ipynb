### Problem 3
Create one dataset with a strong positive correlation and another with a strong negative correlation. Plot them side by side and compare their correlation coefficients.

```python
TimeGamePlayed = np.linspace(0, 4, 8)
QualityOfGameplay = [1, 2, 3, 4, 5, 6, 7, 0]
Dataset1 = pd.DataFrame({
    'Time of Game Play' : TimeGamePlayed,
    'Quality of Game Play' : QualityOfGameplay
})


Number = [0, 10, 20, 30, 40, 50]
Feel = [0, 10, 20, -5, -20, -30]
Dataset2 = pd.DataFrame({
    "Number of Nuggets Eaten": Number,
    "How good I feel" : Feel
})

Dataset1.corr()
fig, axes = plt.subplots(1,2, figsize = (10,5))
axes[0].scatter(Dataset1['Time of Game Play'], Dataset1["Quality of Game Play"], color = 'blue')
axes[1].scatter(Dataset2["Number of Nuggets Eaten"], Dataset2["How good I feel"], color = 'red')
axes[0].set_title("Quality of Game Play over time")
axes[1].set_title("How good I feel after I eat n amount of Nuggets")
axes[0].set_xlabel("Hours")
axes[0].set_ylabel("Quality of Gameplay")
axes[1].set_xlabel("Nuggets Consumed")
axes[1].set_ylabel("How good I feel")
```

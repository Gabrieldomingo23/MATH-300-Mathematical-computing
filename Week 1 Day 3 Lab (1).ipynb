## Problem 1

Write a block of code that tests whether a variable `x` is positive, negative, or zero. Specifically, where "Positive", "Negative", and "Zero" are printed out.

````python
x = float(input("Plug in x"))
if (x > 0):
    print("positive")
elif (x < 0):
    print("x is negative")
else:
    print("x is zero")
````

## Problem 2

Write a block of code that prints out grades. Specifically, given a number `grade`, do the following:

- 'A' if `grade ≥ 90`
- 'B' if `grade ≥ 80`
- 'C' if `grade ≥ 70`
- 'D' if `grade ≥ 60`
- 'F' otherwise

````python
grade = float(input("enter grade"))

if (grade >= 90):
    print("Your grade is an A")
elif grade >= 80:
    print("Your grade is a B")
elif grade >= 70:
    print("Your grade is a C")
elif grade >= 60:
    print("Your grade is a D")
else:
    print("your grade is a F")
````

## Problem 3

A "leap year" is a calendar year where an extra day is added. The rules are as follows:
- If divisible by 400, then leap year
- If divisible by 100, then not a leap year
- If divisible by 4, then leap year
- Otherwise, not a leap year

Write a block of code that takes in the variable `year` and determines if it is a leap year or not.

````python
year = int(input("enter the year"))
if year % 4 == 0:
    print(year, "is a leap year")
elif year % 100 == 0:
    print(year, "is not a leap year")
else:
    print(year, "is not a leap year")
````

## Problem 5

Write a for-loop that keeps a list of the sum of the first `n` integers as `n` ranges from 1 to 30.

````python
sum = 0
for n in range(30):
    sum += n+1
    print(sum)
````
